```python
import numpy as np

arr = np.array([1, 3, 8, 2])
print('numpy_arr:', arr)
```

```python
arr2 = np.array([[1, 3, 8, 2], [2, 6, 7, 5]])
print('numpy_arr:', arr2)

arr2.shape
```

```python
# dtype()
a = np.array([1, 2, 3])
print(a.dtype)
```

```python
# reshape()
a = np.array([1, 2, 3 , 4, 5 ,6])
b = a.reshape(2,3)
print(b)
```

```python
# astype()
a = np.array([1, 2, 3])
b = a.astype(float)
print(b)
```

```python
# np.zeros(), np.ones() dummy data
z = np.zeros((2, 3))
o = np.ones((2, 3))
print("zeros;", z)
print("ones;", o)
```

```python
# ndim
a = np.array([[1, 2], [3, 4]])
print("ones;", a.ndim)
```

```python
# arange()
a = np.arange(1, 6)
print("arrange;", a)
```

```python
# sum(), mean(), min(), max()
sum = np.array([1, 2, 3, 4])
print("sum:", a.sum())
print("mean:", a.mean())
print("min:", a.min())
print("max:", a.max())
```

```python
# flatten()
a = np.array([[1, 2], [3, 4]])
print("flatten;", a.flatten())
```

```python
# T (Transpose)
a = np.array([[1, 2], [3, 4]])
print("Transpose;", a.T)
```

```python
# dot() @
a = np.array([[1, 2], [3, 4]])
b = np.array([[1, 3], [3, 5]])

print("dot;", a.dot(b))
```

```python
# random.rand
a = np.random.rand(2, 3)  # => 0.0 , 1.0   0.0(included)  1.0 (excluded)
print("random.rand;", a)
```
