```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("Student_performance_data _.csv")

print("Dataset shape (rows, columns):", df.shape)
print("Column name:", df.columns.tolist())

print("First 5 rows of the dataset:")
print( df.head())

print("Summary statistics:")
print(df.describe())

print("Data types and null info:")
print(df.info())
```

```python
print("Avarage score by columns:")
df[["Gender", "Ethnicity", "Tutoring", "GPA"]].mean()
```

```python
print("Maximum score by columns:")
df[["Gender", "Ethnicity", "Tutoring", "GPA"]].max()
```

```python
print("Minimum score by columns:")
df[["Gender", "Ethnicity", "Tutoring", "GPA"]].min()
```

```python
# Correlation
print("Correlation:")
df[["Gender", "Ethnicity", "Tutoring", "GPA"]].corr()
```

```python
df["total score"] = df[["Gender", "Ethnicity", "Tutoring", "GPA"]].sum(axis=1)
df["average score"] = df["total score"] / 4
df["passed"] = np.where(df["average score"] >= 50, "Yes", "No")

print("New columns preview (total, average, passed):")
print(df[["total score", "average score", "passed"]].head())
```

```python
test_prep_group = df.groupby("Gender")[['StudentID', 'Age', 'average score']].mean()
print("student score by test preparation course:")
print(test_prep_group)
```

```python
# Gemder-based average
gender_avg = df.groupby("Gender")[['StudentID', 'Age', 'average score']].mean()
print("Average score by gender:")
print(gender_avg)

# Who passed vs failed
pass_rate = df["passed"].value_counts(normalize=True) * 100
print("Pass rate:\n", pass_rate)
```

```python
# Convert scores to Numpy array
scores = df[['StudentID', 'Age', 'average score']].values

# Mean, Std Deviation
print("Numpy Mean:", np.mean(scores, axis=0))
print("Numpy Std:", np.std(scores, axis=0))

# Z-score (normalize)
z_score = (scores - np.mean(scores, axis=0)) / np.std(scores, axis=0)
print("Numpy Z-score:\n", z_score[:5])
```

```python
# How many students scored above 90 average?
top_scorers = df[df["average score"] >=50]
print("Top scored stiudent above 50:", len(top_scorers))

# How many students failed?
failed = df[df["passed"] == "No"]
print("Faied from exam:", len(failed))
```
